```python
import ifcopenshell
from ifccsv import IfcCsv
```

```python
model = ifcopenshell.open("../models/AC20-FZK-Haus.ifc")

elements = model.by_type("IfcWall")
attributes = ["Name", "Description", "OwnerHistory", "ObjectType", "Tag", "PredefinedType", "OverallHeight", "OverallWidth", "LoadBearing", "FireRating", "IsExternal", "BaseQuantities", "Material", "MaterialLayerSet", "Representation"]
```

```python
# Export our model's elements and their attributes to a CSV.
ifc_csv = IfcCsv()

ifc_csv.export(model, elements, attributes, output="../data/ifccsv_example.csv", format="csv", delimiter=",", null="-")
```

```python
df = ifc_csv.export_pd()

df
```
