```python
import ifcopenshell
model = ifcopenshell.open("../models/AC20-FZK-Haus.ifc")
```

```python
print(model.schema)
```

```python
print(model.by_id(66))
```

```python
print(model.by_guid('2XPyKWY018sA1ygZKgQPtU'))
```

```python
print("Number of walls:", len(model.by_type("IfcWall")))
```

```python
wall = model.by_type('IfcWall')[0]
print(wall.is_a())
```

```python
print(wall.is_a('IfcWall')) # Returns True
print(wall.is_a('IfcElement')) # Returns True
print(wall.is_a('IfcWindow')) # Returns False
```

```python
print(wall[0]) # The first attribute is the GlobalId
print(wall[2]) # The third attribute is the Name
```

```python
print(wall.GlobalId)
print(wall.Name)
```

```python
print(wall.id())
```

```python
# Gives us a dictionary of attributes, such as:
# {'id': 8, 'type': 'IfcWall', 'GlobalId': '2_qMTAIHrEYu0vYcqK8cBX', ... }
print(wall.get_info())
```

## Let’s see all the properties and quantities associated with this wall:

```python
import ifcopenshell.util
import ifcopenshell.util.element
print(ifcopenshell.util.element.get_psets(wall))

wall_psets = ifcopenshell.util.element.get_psets(wall)
```

```python
import json

print(json.dumps(ifcopenshell.util.element.get_psets(wall), indent=4))
```

```python
inverse_attr = model.get_inverse(wall)

print(type(inverse_attr))

print(inverse_attr)
```

Let’s do the opposite, let’s see all the elements which our wall references instead:

```python
print(model.traverse(wall))
# Or, let's just go down one level deep
print(model.traverse(wall, max_levels=1))
```

## Changing data

```python
print(wall.Name)

wall.Name = "New Wall Name"
print(wall.Name)
```

## Saving and Creating Files

```python
# You can also generate a new GlobalId:
print(wall.GlobalId)

wall.GlobalId = ifcopenshell.guid.new()
print(wall.GlobalId)
```

```python
# After modifying some IFC data, you can save it to a new IFC-SPF file:

model.write('../models/new.ifc')
```

```python
# You can generate a new IFC from scratch too, instead of reading an existing one:
# ifc = ifcopenshell.file()

# Or if you want a particular schema:
ifc = ifcopenshell.file(schema='IFC4')

ifc.write('../models/empty_model.ifc')
```

## Adding Elements to a model

```python
ifc.create_entity('IfcWall', GlobalId=ifcopenshell.guid.new(), Name='New Wall')

ifc.write('../models/model_with_wall.ifc')
```

```python
# Using a dictionary to create an entity

data = {
    'GlobalId': ifcopenshell.guid.new(),
    'Name': 'Another Wall'
}

ifc.create_entity('IfcWall', **data)

ifc.write('../models/model_with_wall.ifc')
```
